```python
import torch

# Device.
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

# Cuda

```python
# wait until queued GPU operations are ended.
torch.cuda.synchronize(device)

# Memory measure.
mib = 1024**2
memory_allocated = torch.cuda.memory_allocated(device) / mib, 'MiB'
memory_reserved  = torch.cuda.memory_reserved(device) / mib, 'MiB'

torch.cuda.reset_peak_memory_stats(device)              # reset measuring peak.
peak = torch.cuda.max_memory_allocated(device) / mib    # peak since reset.
```

# Tensor

```python
size = (3, 3)

# torch.empty() : uninitiallized values.
x = torch.empty(
    size=size,
    dtype=torch.float32,
    device=device,
    requires_grad=False,
    pin_memory=False,       # work only for CPU tensors.
)
print(f"torch.empty()\n{x}\n")

# torch.zeros(): zeros
x = torch.zeros(size)
print(f"torch.zeros()\n{x}\n")

# torch.ones(): ones.
x = torch.ones(size)
print(f"torch.ones()\n{x}\n")

# torch.triu(): left-bottom=0, upper-right=1.
x = torch.triu(
    torch.ones(size),
    diagonal=1,     # 0 = exact diagonal.
)
print(f"torch.triu()\n{x}\n")

# torch.tril(): left-bottom=0, upper-right=1.
x = torch.tril(
    torch.ones(size),
    diagonal=0,
)
print(f"torch.tril()\n{x}\n")
```

# Random numbers

```python
size = (5, 5)
x = torch.randn(size)
print(f"randn()\n{x}\n")

x = torch.randint(
    low=0,
    high=10,
    size=size,
    dtype=torch.long,
    device=device,
)
print(f"randint()\n{x}\n")
```
