# Setup

```python
from pprint import pprint

# Seed.
seed = 42
```

# 1. Data Preparation

## 1) Load

### from local csv

```python
from datasets import load_dataset, DatasetDict

data = load_dataset(
    path='csv',
    data_files='../tmp/data/titanic/train.csv',
    split='train',  # DatasetDict -> Dataset
)
```

## 2) Split

```python
from datasets import ClassLabel

# Convert label column to ClassLabel, for `stratify_by_column`.
data = data.cast_column(
    'Survived',
    ClassLabel(names=['0', '1']),
)

# Train-valid split.
split = data.train_test_split(
    test_size=0.2,
    shuffle=True,
    stratify_by_column='Survived',  # label name.
    seed=seed,
    keep_in_memory=False,
)

ds = DatasetDict({
    'train': split['train'],
    'valid': split['test'],
})

ds
```
