# 1. Kaggle

## 1) Authentification

- `<path>/access_token`: api key file (without extension)

```python
import os
from pathlib import Path
from kaggle.api.kaggle_api_extended import KaggleApi

# kagglehub.
token_path = Path.home() / ".kaggle" / "access_token"
os.environ["KAGGLE_API_TOKEN"] = token_path.read_text().strip()

# kaggle.
api = KaggleApi()
api.authenticate()
```

## 2) Competition List

```python
# Get competition list.
comps = []

for page in range(1, 6):
    print(f"Page {page}")
    
    resp = api.competitions_list(
        page=page,          # 20 comps per page.
        category='all',     # including closed competitions.
    )
    comps.extend(resp.competitions)

# Attributes for each comp.
for att in dir(comps[0]):
    print(att)

print()

# Sample.
for comp in comps:
    comp = comp.to_dict()
    print(f"- Title: {comp['title']}")
    print(f"  - url: {comp['url']}")
    print(f"  - Description: {comp['description']}")
    print(f"  - Organization: {comp['organizationName']}")
    print(f"  - Start: {comp['dateCreated']}")
    print(f"  - Deadline: {comp['deadline']}")
    break
```
